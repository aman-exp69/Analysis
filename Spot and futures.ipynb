## Analysis to be done

### 🔹 1. Descriptive Statistics
Get a basic sense of their distributions and summary metrics:
- **Mean, Median**
- **Standard Deviation**
- **Skewness & Kurtosis**
- **Min/Max values**
  
**Summary Table:** `pd.describe()`

### 🔹 2. Time Series Visualization
Plot both series together:
- **Line Plot** of both time series
- **Difference plot:** synthetic - actual
- **Rolling mean & standard deviation** (moving window)
- **Histogram** or **KDE plot** of the difference

📌 Visually check for:
- Divergence
- Structural breaks
- Mean-reverting behavior

### 🔹 3. Correlation Checks
- **Pearson Correlation:** Measures linear correlation
- **Rolling Correlation:** To check stability over time
- **Spearman Correlation:** Rank-based (robust to non-linear monotonic relationships)

📌 High correlation indicates similar directional movement.

### 🔹 4. Error Metrics (Regression-Free)
Assess how far apart they are numerically:
- **Mean Absolute Error (MAE)**
- **Mean Squared Error (MSE)**
- **Root Mean Squared Error (RMSE)**
- **Mean Absolute Percentage Error (MAPE)**
- **Maximum Error**

📌 Smaller errors indicate better synthetic replication.

### 🔹 5. Cointegration Test (Statistical Arbitrage Check)
If both series move together but may not be identical, cointegration can confirm long-term equilibrium:
- **Engle-Granger Test**

📌 If p-value < 0.05, they are cointegrated → good synthetic match.

### 🔹 6. Stationarity Check
To compare differences (spread) meaningfully, check:
- **ADF Test** on (synthetic - actual) spread

📌 If spread is stationary, the synthetic future closely tracks the real future over time.

### 🔹 7. Cointegration with Error Correction Model (ECM)
If cointegrated, you can go further by building an **ECM** to track error corrections in the spread.

## Loading Data

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from scipy.stats import gaussian_kde
```

```python
futures_dir = 'Futures'
spot_dir = 'Spot'

def merge_futures_spot_data(futures_dir, spot_dir):
    combined_data = []

    for futures_file in os.listdir(futures_dir):
        if futures_file.endswith('.csv'):
            date_str = futures_file.split('_')[1]
            spot_file = f'NIFTY_{date_str}_Intraday_Spot.csv'
            spot_file_path = os.path.join(spot_dir, spot_file)
            
            if os.path.exists(spot_file_path):
                futures_data = pd.read_csv(os.path.join(futures_dir, futures_file))
                spot_data = pd.read_csv(spot_file_path)
                
                futures_data['Timestamp'] = pd.to_datetime(futures_data['Timestamp'])
                spot_data['Date Time'] = pd.to_datetime(spot_data['Date Time'])
                
                merged_data = pd.merge(futures_data, spot_data, left_on='Timestamp', right_on='Date Time', how='inner')
                
                merged_data = merged_data[['Timestamp', 'Spot', 'mid_price']]
                merged_data.columns = ['Timestamp', 'Spot', 'mid_price']
                merged_data.set_index('Timestamp', inplace=True)
                
                combined_data.append(merged_data)
            else:
                print(f"Warning: No corresponding spot file for {futures_file}")

    final_df = pd.concat(combined_data)
    
    return final_df
```

```python
spot_and_futures_data = merge_futures_spot_data(futures_dir, spot_dir)
spot_and_futures_data = spot_and_futures_data.sort_index(ascending=True)
spot_and_futures_data.rename(columns={'mid_price': 'Futures'}, inplace=True)
spot_and_futures_data = spot_and_futures_data.dropna(subset=['Spot', 'Futures']).copy()
spot_and_futures_data
```

## Spot and Futures Plot & Distribution

```python
def distribution_plots_side_by_side(df, col1, col2, bin_size=60, width=1200, height=600, bar_colour="#99ebff"):

    fig = make_subplots(rows=1, cols=2, subplot_titles=[f"{col1} Distribution", f"{col2} Distribution"])

    for i, col in enumerate([col1, col2], start=1):
        data = df[col].dropna()
        mean = data.mean()
        std = data.std()
        posi_1_std = mean + std
        neg_1_std = mean - std
        posi_2_std = mean + 2*std
        neg_2_std = mean - 2*std

        fig.add_trace(
            go.Histogram(
                x=data,
                nbinsx=bin_size,
                marker_color=bar_colour,
                name=f'{col} Histogram',
                showlegend=(i==1)
            ),
            row=1, col=i
        )

        kde = gaussian_kde(data)
        x_range = np.linspace(data.min(), data.max(), 1000)
        kde_vals = kde(x_range)
        bin_width = (data.max() - data.min()) / bin_size
        scaled_kde_vals = kde_vals * len(data) * bin_width

        fig.add_trace(
            go.Scatter(
                x=x_range,
                y=scaled_kde_vals,
                mode='lines',
                line=dict(color='orange', width=3),
                name=f'{col} KDE',
                showlegend=(i==1)
            ),
            row=1, col=i
        )

        for x_val in [neg_2_std, neg_1_std, mean, posi_1_std, posi_2_std]:
            fig.add_vline(
                x=x_val,
                line_width=2,
                line_dash="dash",
                line_color="red",
                row=1, col=i
            )

        annotations = [
            dict(x=mean, y=max(scaled_kde_vals)*0.9, text=f"Mean(μ)={mean:.2f}", showarrow=True, arrowhead=1),
            dict(x=posi_1_std, y=max(scaled_kde_vals)*0.75, text=f"μ+σ={posi_1_std:.2f}", showarrow=True, arrowhead=1),
            dict(x=neg_1_std, y=max(scaled_kde_vals)*0.75, text=f"μ-σ={neg_1_std:.2f}", showarrow=True, arrowhead=1),
            dict(x=posi_2_std, y=max(scaled_kde_vals)*0.6, text=f"μ+2σ={posi_2_std:.2f}", showarrow=True, arrowhead=1),
            dict(x=neg_2_std, y=max(scaled_kde_vals)*0.6, text=f"μ-2σ={neg_2_std:.2f}", showarrow=True, arrowhead=1),
        ]
        for ann in annotations:
            ann['xref'] = f'x{i}'
            ann['yref'] = f'y{i}'
            fig.add_annotation(ann)

    fig.update_layout(
        title_text="Spot and Futures Distributions",
        font=dict(family="Courier New, monospace", color="RebeccaPurple"),
        width=width,
        height=height,
        paper_bgcolor='white',
        template="plotly_white",
        showlegend=True,
        legend=dict(x=1.05, y=1.0, bordercolor='RebeccaPurple', borderwidth=1)
    )

    fig.show()


distribution_plots_side_by_side(spot_and_futures_data, 'Spot', 'Futures', bin_size=60)
```

## Stationarity Test for Basis Points

```python
spot_and_futures_data['Basis_points'] = ((spot_and_futures_data['Spot'] - spot_and_futures_data['Futures']) / spot_and_futures_data['Futures']) * 10000
basis_points = spot_and_futures_data['Basis_points'].dropna()
```

```python
from statsmodels.tsa.stattools import adfuller

adf_result = adfuller(basis_points)

print("ADF Statistic:", adf_result[0])
print("p-value:", adf_result[1])
print("Critical Values:")
for key, value in adf_result[4].items():
    print(f"   {key}: {value}")
```

p-value << 0.05 <br>
Thus, reject null hypothesis and check plot, and conclude stationarity.

## Conitegration Analysis for Spot and Futures

```python
from statsmodels.tsa.stattools import coint
df = spot_and_futures_data[['Spot', 'Futures']].dropna()
coint_result = coint(df['Spot'], df['Futures'])

coint_stat, p_value, crit_values = coint_result

print("Cointegration Test Statistic:", coint_stat)
print("p-value:", p_value)
print("Critical Values:")
for level, value in zip(['1%', '5%', '10%'], crit_values):
    print(f"   {level}: {value}")
```

```python
def distribution_plots(df, value_col: pd.Series, title_str: str = '', yaxis_title: str = '', xaxis_title: str = ")", width=1000, height=700, bar_colour="#99ebff", bin_size=60):
    mean = value_col.mean()
    median = value_col.median()
    posi_1_std = mean + (1 * value_col.std())
    neg_1_std = mean - (1 * value_col.std())
    posi_2_std = mean + (2 * value_col.std())
    neg_2_std = mean - (2 * value_col.std())

    fig = go.Figure(data=[go.Histogram(x=value_col, nbinsx=bin_size, marker_color=bar_colour)])

    fig.add_vline(x=posi_1_std, line_width=2, line_dash="dash", line_color="red")
    fig.add_vline(x=neg_1_std, line_width=2, line_dash="dash", line_color="red")
    fig.add_vline(x=posi_2_std, line_width=2, line_dash="dash", line_color="red")
    fig.add_vline(x=neg_2_std, line_width=2, line_dash="dash", line_color="red")
    fig.add_vline(x=mean, line_width=2, line_dash="dash", line_color="red")

    fig.add_annotation(x=mean, text=f"Mean(μ)={round(mean, 2)}", ax=80, ay=-150, showarrow=True, font=dict(size=15), arrowhead=1)
    fig.add_annotation(x=posi_1_std, text=f"μ+σ={round(posi_1_std, 2)}", ax=80, ay=-80, showarrow=True, font=dict(size=15), arrowhead=1)
    fig.add_annotation(x=neg_1_std, text=f"μ-σ={round(neg_1_std, 2)}", ax=-80, ay=-80, showarrow=True, font=dict(size=15), arrowhead=1)
    fig.add_annotation(x=posi_2_std, text=f"μ+2σ={round(posi_2_std, 2)}", ax=80, ay=-60, showarrow=True, font=dict(size=15), arrowhead=1)
    fig.add_annotation(x=neg_2_std, text=f"μ-2σ={round(neg_2_std, 2)}", ax=-80, ay=-60, showarrow=True, font=dict(size=15), arrowhead=1)

    fig.update_layout(
        title=title_str,
        yaxis=dict(title=yaxis_title, showline=True, linecolor='RebeccaPurple', linewidth=1, mirror=True),
        xaxis=dict(title=xaxis_title, showline=True, linecolor='RebeccaPurple', linewidth=1, mirror=True),
        font=dict(family="Courier New, monospace", color="RebeccaPurple"),
        width=width,
        height=height,
        legend=dict(x=1.0, y=1.0, xanchor='right', yanchor='top', orientation='v', bordercolor='RebeccaPurple', borderwidth=1),
        margin=dict(l=60, r=60, b=60, t=60, pad=5),
        paper_bgcolor='white',
        template="plotly_white"
    )

    fig.show()
```

## Analysis for Basis Points

```python
spot_and_futures_data = merge_futures_spot_data(futures_dir, spot_dir)
spot_and_futures_data = spot_and_futures_data.sort_index(ascending=True)
spot_and_futures_data.rename(columns={'mid_price': 'Futures'}, inplace=True)
spot_and_futures_data = spot_and_futures_data.dropna(subset=['Spot', 'Futures']).copy()
spot_and_futures_data
```

```python
def compute_basis_points(df):
    return ((df['Spot'] - df['Futures']) / df['Futures']) * 10000

time_intervals = [
    ('09:16', '10:30'),
    ('10:30', '11:30'),
    ('11:30', '12:30'),
    ('12:30', '13:30'),
    ('13:30', '14:30'),
    ('14:30', '15:30')
]

df = spot_and_futures_data.copy()
df.index = pd.to_datetime(df.index)

results = []

for date, group in df.groupby(df.index.date):
    for start_time, end_time in time_intervals:
        start = pd.Timestamp(f"{date} {start_time}")
        end = pd.Timestamp(f"{date} {end_time}")
        
        interval_df = group[(group.index >= start) & (group.index < end)]
        
        if not interval_df.empty:
            basis_points = compute_basis_points(interval_df)
            
            open_bp = basis_points.iloc[0]
            close_bp = basis_points.iloc[-1]
            
            if close_bp > open_bp:
                direction = 'Increase'
            elif close_bp < open_bp:
                direction = 'Decrease'
            else:
                direction = 'Flat'

            results.append({
                'Date': pd.to_datetime(date),
                'Interval': f"{start_time}-{end_time}",
                'AvgBasisPoint': basis_points.mean(),
                'StdBasisPoint': basis_points.std(),
                'MinBasisPoint': basis_points.min(),
                'MaxBasisPoint': basis_points.max(),
                'OpenBasisPoint': open_bp,
                'CloseBasisPoint': close_bp,
                'ChangeDirection': direction
            })

interval_stats_df = pd.DataFrame(results)
```
